## Task 04: Sets for Unique Vocabulary

```python
# Task 04 (a)
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]
vocab_set = set(raw)
print(f"Raw list: {raw}")
print(f"Vocab set: {vocab_set}")
print(f"Number of unique words: {len(vocab_set)}")
# Sets automatically remove duplicates because they use hashing; each element must be unique to have a single hash key entry.

# Task 04 (b)
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}
union_set = set_A | set_B
intersection_set = set_A & set_B
difference_set = set_A - set_B
print(f"\nUnion: {union_set}")
print(f"Intersection: {intersection_set}")
print(f"Difference (A - B): {difference_set}")

# Task 04 (c)
stopwords = {"is", "the", "a", "in", "of", "and", "to"}
tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]
cleaned_set = set(tokens) - stopwords
# Note: set difference removes duplicates and original order
print(f"\nCleaned tokens using set difference: {cleaned_set}")

# Task 04 (d)
vocab_set.update(["transformer", "bert"])
vocab_set.discard("python")
has_ai = "ai" in vocab_set
print(f"\n'ai' is still present: {has_ai}")
print(f"Final vocab set: {vocab_set}")
```

## Task 06: String Cleaning Methods

```python
# Task 06 (a)
text = "   MACHINE Learning is AMAZING!!!   "
step1 = text.strip()
print(f"After strip(): '{step1}'")
step2 = step1.lower()
print(f"After lower(): '{step2}'")
step3 = step2.replace("!", "")
print(f"After replace(): '{step3}'")

# Task 06 (b)
cleaned_text = step3
tokenised_list = cleaned_text.split()
print(f"\nTokenised list: {tokenised_list}")
uppercase_list = [word.upper() for word in tokenised_list]
print(f"Uppercase list: {uppercase_list}")
joined_string = " | ".join(uppercase_list)
print(f"Joined string: {joined_string}")

# Task 06 (c)
sentence = "I love Java and Java is popular"
updated_sentence = sentence.replace("Java", "Python")
print(f"\nUpdated sentence: {updated_sentence}")
python_pos = updated_sentence.find("Python")
print(f"Position of first 'Python': {python_pos}")

# Task 06 (d)
def normalize(text):
    return text.strip().lower().replace(",", "").replace(".", "")

test_str1 = "  Hello, World. This is NLP.  "
test_str2 = "MACHINE, Learning, is, FUN."
print(f"\nNormalized string 1: '{normalize(test_str1)}'")
print(f"Normalized string 2: '{normalize(test_str2)}'")
```

## Task 07: String Validation Methods

```python
# Task 07 (a)
text = "NLP is the future of AI research"
starts_with_nlp = text.startswith("NLP")
ends_with_research = text.endswith("research")
print(f"Text begins with 'NLP': {starts_with_nlp}")
print(f"Text ends with 'research': {ends_with_research}")
future_pos = text.find("future")
print(f"Position of 'future': {future_pos}")

# Task 07 (b)
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
print("\nToken types:")
for token in tokens:
    if token.isalpha():
        print(f"'{token}': Purely alphabetic")
    elif token.isdigit():
        print(f"'{token}': Purely numeric")
    else:
        print(f"'{token}': Mixed")

# Task 07 (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print(f"\nPure words: {pure_words}")
print(f"Pure numbers: {pure_numbers}")
print(f"Mixed tokens: {mixed_tokens}")

# Task 07 (d)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
print("\nValidated tokens:")
for t in test_tokens:
    print(f"'{t}': {validate_token(t)}")
```

## Task 08: Functions and Lambda for NLP Pipelines

```python
# Task 08 (a)
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

test_text = "  HELLO Natural Language Processing  "
print(f"clean() output: '{clean(test_text)}'")
print(f"clean_lambda output: '{clean_lambda(test_text)}'")

# Task 08 (b)
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
cleaned_sentence = clean(sentence)
tokens_list = tokenize(cleaned_sentence)
filtered_tokens = remove_short(tokens_list, 3)
print(f"\nFiltered tokens (len > 3): {filtered_tokens}")

# Task 08 (c)
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

sample_tokens = ["language", "model", "ai", "processing", "deep"]
print(f"\nWord lengths dictionary: {word_lengths(sample_tokens)}")

# Task 08 (d)
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

sentence1 = "Artificial intelligence is amazing and fun"
sentence2 = "Deep learning helps machines understand text"

print(f"\nPipeline result 1: {nlp_pipeline(sentence1, 3)}")
print(f"Pipeline result 2: {nlp_pipeline(sentence2, 4)}")
```

## Task 09: map(), filter(), sorted() and Counter

```python
# Task 09 (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]
lowercase_words = list(map(lambda x: x.lower(), words))
word_lengths_list = list(map(lambda x: len(x), words))
print(f"Lowercase words: {lowercase_words}")
print(f"Word lengths: {word_lengths_list}")

# Task 09 (b)
words_long = list(filter(lambda x: len(x) > 3, words))
words_with_t = list(filter(lambda x: 't' in x.lower(), words))
print(f"\nWords longer than 3 characters: {words_long}")
print(f"Words containing 't': {words_with_t}")

# Task 09 (c)
sorted_alphabetically = sorted(words)
sorted_len_asc = sorted(words, key=len)
sorted_len_desc = sorted(words, key=len, reverse=True)
print(f"\nSorted alphabetically: {sorted_alphabetically}")
print(f"Sorted by length (ascending): {sorted_len_asc}")
print(f"Sorted by length (descending): {sorted_len_desc}")

# Task 09 (d)
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
words_split = sentence.split()
freq = Counter(words_split)

print(f"\nFull Counter: {freq}")
print(f"3 most common words: {freq.most_common(3)}")
print(f"Frequency of 'python': {freq['python']}")
print(f"Total number of words: {sum(freq.values())}")
print(f"Size of unique vocabulary: {len(freq)}")
```
