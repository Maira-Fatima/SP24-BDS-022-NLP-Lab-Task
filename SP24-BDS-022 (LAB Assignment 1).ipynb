```python
import re
import nltk
import spacy
from nltk.tokenize import word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

## *Task 1: Regex Fundamentals (Character Classes & Quantifiers)*

```python
text1 = "In 2023, Bilal scored 95 marks in Data Science. Ali and Sara bought 12 books for Rs. 500 in Lahore during 1999 and 2024! Total: 3 pens & 7 cups."

# a) all digits (each single digit)
digits = re.findall(r"\d", text1)
print("a) Digits:", digits)

# b) words starting with a capital letter
capital_words = re.findall(r"\b[A-Z][a-zA-Z]*\b", text1)
print("b) Capitalized words:", capital_words)

# c) sequences of exactly 4 digits (lookarounds stop matching inside longer numbers)
four_digits = re.findall(r"(?<!\d)\d{4}(?!\d)", text1)
print("c) Exactly 4 digits:", four_digits)

# d) words 3 to 6 letters long
words_3_6 = re.findall(r"\b[a-zA-Z]{3,6}\b", text1)
print("d) Words of length 3-6:", words_3_6)
```

## *Task 2: Pattern-Based Extraction (Real World Data)*

```python
para = """Please contact support@shop.com or sales.team@my-store.co.pk for help.
Call 0300-1234567 or 0345-7654321 before 15/08/2025. Order IDs ORD-4521 and ORD-9987
were placed on 01/09/2025 and 23/12/2024. The discount was 45% for the first, and 12.5% for the second.
Late fee is 3.75% per month. Ignore 1234-567 and 32/13/25 (invalid formats)."""

# a) emails
emails = re.findall(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", para)
print("a) Emails:", emails)

# b) phone numbers: 03XX-XXXXXXX
phones = re.findall(r"\b03\d{2}-\d{7}\b", para)
print("b) Phones:", phones)

# c) dates DD/MM/YYYY (format check only, does not validate real calendar dates)
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", para)
print("c) Dates:", dates)

# d) percentages, integer or decimal
percents = re.findall(r"\b\d+(?:\.\d+)?%", para)
print("d) Percentages:", percents)
```

## *Task 3: Word Tokenization (NLTK vs spaCy)*

```python
paragraph = ("I don't think it's easy to learn NLP in 2024. Dr. Khan said the well-known U.S. course costs $50, "
             "and \"practice\" matters! We can't skip e-mail parsing, can we?")

# a) NLTK
nltk_tokens = word_tokenize(paragraph)
print("NLTK :", nltk_tokens)

# b) blank spaCy pipeline (tokenizer only)
nlp_blank = spacy.blank("en")
spacy_tokens = [t.text for t in nlp_blank(paragraph)]
print("spaCy:", spacy_tokens)

# d) token counts
print("\nNLTK token count :", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))

# c) Differences (noted in the comment below)
# 1. Hyphenated words: NLTK keeps "well-known" and "e-mail" as single tokens, while spaCy splits them
#    into "well", "-", "known" and "e", "-", "mail" (spaCy splits infix hyphens).
# 2. Quotation marks: NLTK converts double quotes into `` and '' (Penn Treebank style),
#    while spaCy keeps the original " character as its own token.
# 3. Both split contractions the same way ("do" + "n't", "it" + "'s", "ca" + "n't"),
#    and both keep "Dr." and "U.S." intact. spaCy gives more tokens overall because of the hyphen splits.
```

## *Task 4: Token Attributes (Categorizing Tokens)*

```python
sentence = "Contact me at bilal@example.com to pay $50 or Rs. 500 for 3 books."
doc = nlp_blank(sentence)   # a) Doc object

print(f"{'TOKEN':<20}{'is_alpha':<10}{'like_num':<10}{'like_email':<12}")
for token in doc:
    # b) is_alpha   c) like_num and like_email
    print(f"{token.text:<20}{str(token.is_alpha):<10}{str(token.like_num):<10}{str(token.like_email):<12}")

# d) only currency tokens
currency_tokens = [t.text for t in doc if t.is_currency]
print("\nCurrency tokens:", currency_tokens)
```

Note: spaCy's `is_currency` flags symbols like `$`. The text `Rs` is treated as a normal alphabetic word, so it is not flagged.

## *Task 5: Sentence Segmentation (Handling Tricky Cases)*

```python
# a) trained pipeline
nlp = spacy.load("en_core_web_sm")

text5 = ("Dr. Smith met Mr. Ali at 5 p.m. on Monday. They discussed AI, e.g. chatbots and agents. "
         "It went well.")
doc5 = nlp(text5)

# b) split into sentences
sents = list(doc5.sents)
for i, s in enumerate(sents, 1):
    print(f"{i}. {s.text}")

# c) total sentences
print("\nTotal sentences:", len(sents))

# d) did any abbreviation cause a wrong break?
abbrevs = ("Dr.", "Mr.", "p.m.", "e.g.")
wrong = [s.text for s in sents if s.text.strip().endswith(abbrevs) and s.text.strip() != sents[-1].text.strip()]
print("Sentences wrongly ended at an abbreviation:", wrong)
```

**Observation:** No incorrect breaks happened. `Dr.`, `Mr.`, `p.m.` and `e.g.` stayed inside their sentences because spaCy's tokenizer treats them as known abbreviations and the dependency parser (which drives `doc.sents` in `en_core_web_sm`) uses full context, not just full stops.
